# Factory Reallocation & Shipping Optimization Recommendation System for Nassau Candy Distributor

## Project Overview

Nassau Candy Distributor currently uses static factory-product assignments, which can result in operational inefficiencies such as long shipping lead times and inconsistent route performance. This project develops a decision intelligence system to analyze existing shipping patterns, predict expected shipping lead times, simulate alternative factory assignments, and generate data-driven factory reallocation recommendations.

The system combines exploratory data analysis, machine learning, route clustering, scenario simulation, and optimization logic. Three regression models — Linear Regression, Random Forest Regressor, and Gradient Boosting Regressor — are evaluated to predict shipping lead time.

The selected predictive model is then used to simulate different factory configurations for products across destination regions and shipping modes. These scenarios are evaluated using lead-time improvement, scenario confidence, and profit stability indicators to generate practical factory reallocation recommendations.

### Key Objectives

- Predict shipping lead time under different shipping configurations.
- Identify product-region combinations with potential factory reallocation opportunities.
- Compare current and alternative factory assignments using what-if simulation.
- Cluster factory-region routes according to performance characteristics.
- Balance shipping efficiency with financial stability.
- Generate scalable recommendations for decision-making through a Streamlit dashboard.

### Business Value

The project moves Nassau Candy Distributor from descriptive analysis toward intelligent decision support. Instead of only identifying historical performance patterns, the system provides a framework for testing alternative factory assignments and quantifying their predicted operational impact before implementation.

## Dataset Description

The analysis uses a transaction-level Nassau Candy Distributor dataset containing 10,194 records and 18 original fields covering orders, customers, geography, products, sales, units, gross profit, cost, shipping mode, order date, and ship date.

For this project, the dataset was enriched with factory assignment information and factory coordinates supplied as part of the project context. Fifteen products are mapped to five factories, while customer destinations are represented through country/region, state/province, city, postal code, postal code, and business region.

> **Scope Note:** Factory reallocation results are model-based what-if estimates. The dataset does not contain historical observations of every product being manufactured at every factory, nor does it provide factory-specific transportation costs. Therefore, the scenario engine estimates operational differences from the predictive model rather than claiming experimentally observed causal effects.

## Problem Statement

Nassau Candy Distributor operates with fixed product-to-factory assignments. While this approach provides a stable operational structure, it may not always provide the most efficient shipping outcomes for every destination and shipping condition.

Different factory locations can create differences in predicted shipping lead time across products, regions, and shipping modes. However, simply moving products between factories may also affect operational and financial stability. Therefore, a systematic approach is required to identify potential reallocation opportunities and compare them using measurable performance indicators.

This project addresses the problem by building a predictive and recommendation-based decision support system. The system analyzes historical transaction data, predicts shipping lead time, evaluates alternative factory assignments through what-if scenarios, and combines operational improvement with profit-stability indicators.

The final system is intended to support informed planning by showing where alternative factory assignments may provide predicted improvements and where the available data does not indicate a positive reallocation opportunity.

## Analytical Assumptions and Scope

The analysis follows several assumptions to maintain consistency between historical analysis, predictive modeling, and factory scenario simulation.

1. The original transaction records are retained at the transaction level. Repeated Order IDs, Customer IDs, and Product IDs are not treated as duplicate observations because they represent legitimate business transactions.

2. `Lead_Time_Days` is calculated directly from the available order and shipment dates. The resulting values are analyzed as provided by the dataset without artificially modifying the date records.

3. The supplied product-to-factory mapping is treated as the current factory configuration for the analysis.

4. Alternative factory scenarios are evaluated using the trained machine learning model. These scenarios represent what-if estimates rather than experimentally observed outcomes.

5. Profit stability is evaluated using available historical profit-related indicators such as profit margin and profit per unit. Factory-specific transportation or manufacturing costs are not available in the dataset and are therefore not included as direct monetary optimization variables.

6. Recommendation scores are analytical decision-support measures created for this project. They are intended to compare scenarios consistently and should not be interpreted as guaranteed operational outcomes.

This framework keeps the analysis transparent while allowing historical performance, predictive modeling, and scenario evaluation to be combined into a single decision-support workflow.

```python
# ============================================================
# PREPARE ROUTE FEATURES FOR CLUSTERING
# ============================================================

from sklearn.preprocessing import StandardScaler

# Performance features selected for clustering
clustering_features = [
    "Average_Lead_Time",
    "Lead_Time_Variability",
    "Average_Profit_Margin",
    "Profit_Per_Record",
    "Sales_Per_Record"
]

# Create clustering dataset
cluster_data = route_features[
    [
        "Current_Factory",
        "Region"
    ] + clustering_features
].copy()

# ------------------------------------------------------------
# Standardize numerical features
# ------------------------------------------------------------

scaler = StandardScaler()

cluster_data_scaled = scaler.fit_transform(
    cluster_data[clustering_features]
)

# Convert scaled values back into a DataFrame
cluster_scaled_df = pd.DataFrame(
    cluster_data_scaled,
    columns=clustering_features,
    index=cluster_data.index
)

print("=" * 70)
print("CLUSTERING DATA PREPARATION")
print("=" * 70)

print("\nFeatures used for clustering:")
for feature in clustering_features:
    print("-", feature)

print("\nNumber of routes:")
print(len(cluster_data))

print("\nScaled feature matrix shape:")
print(cluster_scaled_df.shape)

print("\nScaled clustering features:")
display(cluster_scaled_df.head())
```

```python
# ============================================================
# DETERMINE OPTIMAL NUMBER OF ROUTE CLUSTERS
# ============================================================

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Test cluster counts from 2 to 6
k_values = range(2, 7)

inertia_values = []
silhouette_values = []

for k in k_values:

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(cluster_scaled_df)

    inertia_values.append(kmeans.inertia_)
    silhouette_values.append(
        silhouette_score(
            cluster_scaled_df,
            labels
        )
    )

# ------------------------------------------------------------
# Create comparison table
# ------------------------------------------------------------

cluster_evaluation = pd.DataFrame({
    "Number_of_Clusters": list(k_values),
    "Inertia": inertia_values,
    "Silhouette_Score": silhouette_values
})

print("=" * 70)
print("CLUSTER NUMBER EVALUATION")
print("=" * 70)

display(
    cluster_evaluation.round(4)
)

# ------------------------------------------------------------
# Identify best silhouette score
# ------------------------------------------------------------

best_k = cluster_evaluation.loc[
    cluster_evaluation["Silhouette_Score"].idxmax(),
    "Number_of_Clusters"
]

best_silhouette = cluster_evaluation[
    "Silhouette_Score"
].max()

print("\nBest cluster count based on Silhouette Score:")
print(int(best_k))

print("\nBest Silhouette Score:")
print(round(best_silhouette, 4))

# ------------------------------------------------------------
# Plot Elbow Curve
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    list(k_values),
    inertia_values,
    marker="o"
)

plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")
plt.title("Elbow Method for Route Clustering")
plt.xticks(list(k_values))
plt.grid(True, alpha=0.3)

plt.show()

# ------------------------------------------------------------
# Plot Silhouette Scores
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    list(k_values),
    silhouette_values,
    marker="o"
)

plt.xlabel("Number of Clusters")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score by Number of Clusters")
plt.xticks(list(k_values))
plt.grid(True, alpha=0.3)

plt.show()
```

```python
# ============================================================
# PREPARE DATA FOR LEAD-TIME PREDICTION
# ============================================================

from sklearn.model_selection import train_test_split

# ------------------------------------------------------------
# Select model features
# ------------------------------------------------------------

model_features = [
    "Product_Name",
    "Current_Factory",
    "Region",
    "Ship_Mode",
    "Division",
    "Country_Region",
    "Order_Year",
    "Order_Month",
    "Order_Quarter",
    "Order_DayOfWeek",
    "Is_Weekend",
    "Sales",
    "Units",
    "Gross_Profit",
    "Cost",
    "Profit_Margin",
    "Sales_Per_Unit",
    "Cost_Per_Unit",
    "Profit_Per_Unit"
]

target = "Lead_Time_Days"

# ------------------------------------------------------------
# Create model dataset
# ------------------------------------------------------------

model_df = clean_df[
    model_features + [target]
].copy()

# Remove any accidental missing values
model_df = model_df.dropna()

X = model_df[model_features]
y = model_df[target]

# ------------------------------------------------------------
# Identify feature types
# ------------------------------------------------------------

categorical_features = [
    "Product_Name",
    "Current_Factory",
    "Region",
    "Ship_Mode",
    "Division",
    "Country_Region"
]

numeric_features = [
    "Order_Year",
    "Order_Month",
    "Order_Quarter",
    "Order_DayOfWeek",
    "Is_Weekend",
    "Sales",
    "Units",
    "Gross_Profit",
    "Cost",
    "Profit_Margin",
    "Sales_Per_Unit",
    "Cost_Per_Unit",
    "Profit_Per_Unit"
]

print("=" * 70)
print("LEAD-TIME PREDICTION DATASET")
print("=" * 70)

print("\nModel dataset shape:")
print(model_df.shape)

print("\nTarget variable:")
print(target)

print("\nNumber of features:")
print(len(model_features))

print("\nCategorical features:")
for feature in categorical_features:
    print("-", feature)

print("\nNumerical features:")
for feature in numeric_features:
    print("-", feature)

# ------------------------------------------------------------
# Train-test split
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\n" + "=" * 70)
print("TRAIN-TEST SPLIT")
print("=" * 70)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))

print("\nTarget training shape:", y_train.shape)
print("Target testing shape:", y_test.shape)
```

```python
# ============================================================
# BUILD MACHINE LEARNING PREPROCESSING PIPELINE
# ============================================================

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)

# ------------------------------------------------------------
# Categorical preprocessing
# ------------------------------------------------------------

categorical_transformer = Pipeline(
    steps=[
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

# ------------------------------------------------------------
# Numerical preprocessing
# ------------------------------------------------------------

numeric_transformer = Pipeline(
    steps=[
        (
            "scaler",
            StandardScaler()
        )
    ]
)

# ------------------------------------------------------------
# Combine both preprocessing methods
# ------------------------------------------------------------

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            categorical_transformer,
            categorical_features
        ),
        (
            "numerical",
            numeric_transformer,
            numeric_features
        )
    ]
)

print("=" * 70)
print("PREPROCESSING PIPELINE CREATED")
print("=" * 70)

print("\nCategorical preprocessing:")
print("One-Hot Encoding")

print("\nNumerical preprocessing:")
print("Standard Scaling")

print("\nUnknown categories:")
print("Ignored safely using handle_unknown='ignore'")

print("\nPreprocessing pipeline is ready.")
```

```python
# ============================================================
# MODEL 1 - LINEAR REGRESSION BASELINE
# ============================================================

from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# ------------------------------------------------------------
# Create Linear Regression pipeline
# ------------------------------------------------------------

linear_model = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "model",
            LinearRegression()
        )
    ]
)

# ------------------------------------------------------------
# Train the model
# ------------------------------------------------------------

linear_model.fit(
    X_train,
    y_train
)

# ------------------------------------------------------------
# Generate predictions
# ------------------------------------------------------------

linear_predictions = linear_model.predict(
    X_test
)

# ------------------------------------------------------------
# Evaluate the model
# ------------------------------------------------------------

linear_mae = mean_absolute_error(
    y_test,
    linear_predictions
)

linear_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        linear_predictions
    )
)

linear_r2 = r2_score(
    y_test,
    linear_predictions
)

# ------------------------------------------------------------
# Display results
# ------------------------------------------------------------

linear_results = pd.DataFrame({
    "Model": ["Linear Regression"],
    "MAE": [linear_mae],
    "RMSE": [linear_rmse],
    "R2_Score": [linear_r2]
})

print("=" * 70)
print("LINEAR REGRESSION RESULTS")
print("=" * 70)

display(
    linear_results.round(4)
)
```

```python
# ============================================================
# MODEL 2 - RANDOM FOREST REGRESSOR
# ============================================================

from sklearn.ensemble import RandomForestRegressor

# ------------------------------------------------------------
# Create Random Forest pipeline
# ------------------------------------------------------------

random_forest_model = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "model",
            RandomForestRegressor(
                n_estimators=200,
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)

# ------------------------------------------------------------
# Train the model
# ------------------------------------------------------------

random_forest_model.fit(
    X_train,
    y_train
)

# ------------------------------------------------------------
# Generate predictions
# ------------------------------------------------------------

rf_predictions = random_forest_model.predict(
    X_test
)

# ------------------------------------------------------------
# Evaluate the model
# ------------------------------------------------------------

rf_mae = mean_absolute_error(
    y_test,
    rf_predictions
)

rf_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        rf_predictions
    )
)

rf_r2 = r2_score(
    y_test,
    rf_predictions
)

# ------------------------------------------------------------
# Display results
# ------------------------------------------------------------

rf_results = pd.DataFrame({
    "Model": ["Random Forest Regressor"],
    "MAE": [rf_mae],
    "RMSE": [rf_rmse],
    "R2_Score": [rf_r2]
})

print("=" * 70)
print("RANDOM FOREST REGRESSOR RESULTS")
print("=" * 70)

display(
    rf_results.round(4)
)
```

```python
# ============================================================
# MODEL 3 - GRADIENT BOOSTING REGRESSOR
# ============================================================

from sklearn.ensemble import GradientBoostingRegressor

# ------------------------------------------------------------
# Create Gradient Boosting pipeline
# ------------------------------------------------------------

gradient_boosting_model = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "model",
            GradientBoostingRegressor(
                n_estimators=200,
                learning_rate=0.05,
                max_depth=3,
                random_state=42
            )
        )
    ]
)

# ------------------------------------------------------------
# Train the model
# ------------------------------------------------------------

gradient_boosting_model.fit(
    X_train,
    y_train
)

# ------------------------------------------------------------
# Generate predictions
# ------------------------------------------------------------

gb_predictions = gradient_boosting_model.predict(
    X_test
)

# ------------------------------------------------------------
# Evaluate the model
# ------------------------------------------------------------

gb_mae = mean_absolute_error(
    y_test,
    gb_predictions
)

gb_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        gb_predictions
    )
)

gb_r2 = r2_score(
    y_test,
    gb_predictions
)

# ------------------------------------------------------------
# Display results
# ------------------------------------------------------------

gb_results = pd.DataFrame({
    "Model": ["Gradient Boosting Regressor"],
    "MAE": [gb_mae],
    "RMSE": [gb_rmse],
    "R2_Score": [gb_r2]
})

print("=" * 70)
print("GRADIENT BOOSTING REGRESSOR RESULTS")
print("=" * 70)

display(
    gb_results.round(4)
)
```

```python
# ============================================================
# FACTORY REALLOCATION IMPACT ANALYSIS
# ============================================================

# ------------------------------------------------------------
# Aggregate predicted lead time for each
# Product × Region × Candidate Factory
# ------------------------------------------------------------

scenario_summary = (
    factory_scenario_predictions
    .groupby(
        [
            "Product_ID",
            "Product_Name",
            "Division",
            "Region",
            "Candidate_Factory",
            "Is_Current_Factory"
        ],
        as_index=False
    )
    .agg(
        Records=("Predicted_Lead_Time", "size"),
        Predicted_Lead_Time=("Predicted_Lead_Time", "mean")
    )
)

# Round prediction
scenario_summary["Predicted_Lead_Time"] = (
    scenario_summary["Predicted_Lead_Time"].round(2)
)

# ------------------------------------------------------------
# Get predicted lead time under the current factory
# ------------------------------------------------------------

current_prediction = (
    scenario_summary[
        scenario_summary["Is_Current_Factory"]
    ][
        [
            "Product_ID",
            "Product_Name",
            "Region",
            "Predicted_Lead_Time"
        ]
    ]
    .rename(
        columns={
            "Predicted_Lead_Time":
            "Current_Predicted_Lead_Time"
        }
    )
)

# ------------------------------------------------------------
# Merge current prediction into every scenario
# ------------------------------------------------------------

scenario_summary = scenario_summary.merge(
    current_prediction,
    on=[
        "Product_ID",
        "Product_Name",
        "Region"
    ],
    how="left"
)

# ------------------------------------------------------------
# Calculate expected improvement
# ------------------------------------------------------------

scenario_summary["Lead_Time_Reduction_Days"] = (
    scenario_summary["Current_Predicted_Lead_Time"]
    - scenario_summary["Predicted_Lead_Time"]
)

scenario_summary["Lead_Time_Reduction_Percent"] = np.where(
    scenario_summary["Current_Predicted_Lead_Time"] != 0,
    (
        scenario_summary["Lead_Time_Reduction_Days"]
        / scenario_summary["Current_Predicted_Lead_Time"]
    ) * 100,
    0
)

scenario_summary["Lead_Time_Reduction_Days"] = (
    scenario_summary["Lead_Time_Reduction_Days"].round(2)
)

scenario_summary["Lead_Time_Reduction_Percent"] = (
    scenario_summary["Lead_Time_Reduction_Percent"].round(2)
)

# ------------------------------------------------------------
# Sort by largest predicted improvement
# ------------------------------------------------------------

scenario_summary = (
    scenario_summary
    .sort_values(
        "Lead_Time_Reduction_Days",
        ascending=False
    )
    .reset_index(drop=True)
)

print("=" * 70)
print("FACTORY REALLOCATION IMPACT")
print("=" * 70)

print("\nTotal scenario combinations:")
print(len(scenario_summary))

print("\nTop 25 scenarios by predicted lead-time reduction:")

display(
    scenario_summary[
        [
            "Product_ID",
            "Product_Name",
            "Region",
            "Candidate_Factory",
            "Is_Current_Factory",
            "Current_Predicted_Lead_Time",
            "Predicted_Lead_Time",
            "Lead_Time_Reduction_Days",
            "Lead_Time_Reduction_Percent"
        ]
    ].head(25)
)

# ------------------------------------------------------------
# Save scenario summary
# ------------------------------------------------------------

scenario_output_path = (
    OUTPUT_DIR / "factory_reallocation_scenarios.csv"
)

scenario_summary.to_csv(
    scenario_output_path,
    index=False
)

print("\nScenario analysis saved to:")
print(scenario_output_path)
```

```python
# ============================================================
# FACTORY REALLOCATION RECOMMENDATION ENGINE
# ============================================================

# ------------------------------------------------------------
# Create alternative-factory scenarios
# ------------------------------------------------------------

recommendation_candidates = scenario_summary[
    ~scenario_summary["Is_Current_Factory"]
].copy()

# ------------------------------------------------------------
# Calculate speed score
# ------------------------------------------------------------

# Normalize lead-time reduction relative to the best alternative
max_reduction = recommendation_candidates[
    "Lead_Time_Reduction_Days"
].max()

if max_reduction > 0:
    recommendation_candidates["Speed_Score"] = (
        recommendation_candidates["Lead_Time_Reduction_Days"]
        / max_reduction
    ) * 100
else:
    recommendation_candidates["Speed_Score"] = 0

# ------------------------------------------------------------
# Create scenario confidence proxy
# ------------------------------------------------------------

# More historical product-region records provide stronger
# empirical support for the scenario.
recommendation_candidates["Volume_Confidence"] = np.minimum(
    recommendation_candidates["Records"] / 50,
    1
)

# Model explanatory performance contributes to the confidence proxy.
model_r2 = gb_r2

recommendation_candidates["Model_Confidence"] = model_r2

recommendation_candidates["Scenario_Confidence_Score"] = (
    0.6 * recommendation_candidates["Volume_Confidence"]
    + 0.4 * recommendation_candidates["Model_Confidence"]
) * 100

# ------------------------------------------------------------
# Recommendation score
# ------------------------------------------------------------

recommendation_candidates["Recommendation_Score"] = (
    0.70 * recommendation_candidates["Speed_Score"]
    + 0.30 * recommendation_candidates[
        "Scenario_Confidence_Score"
    ]
)

# ------------------------------------------------------------
# Classify recommendation risk
# ------------------------------------------------------------

recommendation_candidates["Recommendation_Risk"] = np.select(
    [
        recommendation_candidates[
            "Scenario_Confidence_Score"
        ] >= 70,

        recommendation_candidates[
            "Scenario_Confidence_Score"
        ] >= 45
    ],
    [
        "Low",
        "Medium"
    ],
    default="High"
)

# ------------------------------------------------------------
# Rank alternatives for every product-region combination
# ------------------------------------------------------------

recommendation_candidates[
    "Recommendation_Rank"
] = (
    recommendation_candidates
    .groupby(
        [
            "Product_ID",
            "Product_Name",
            "Region"
        ]
    )["Recommendation_Score"]
    .rank(
        ascending=False,
        method="first"
    )
)

# ------------------------------------------------------------
# Select Top-3 recommendations
# ------------------------------------------------------------

top_recommendations = (
    recommendation_candidates[
        recommendation_candidates["Recommendation_Rank"] <= 3
    ]
    .sort_values(
        [
            "Product_Name",
            "Region",
            "Recommendation_Rank"
        ]
    )
    .reset_index(drop=True)
)

print("=" * 70)
print("FACTORY REALLOCATION RECOMMENDATION ENGINE")
print("=" * 70)

print("\nAlternative-factory scenarios evaluated:")
print(len(recommendation_candidates))

print("\nTop recommendations generated:")
print(len(top_recommendations))

print("\n" + "=" * 70)
print("TOP 3 RECOMMENDATIONS PER PRODUCT × REGION")
print("=" * 70)

display(
    top_recommendations[
        [
            "Product_ID",
            "Product_Name",
            "Region",
            "Candidate_Factory",
            "Current_Predicted_Lead_Time",
            "Predicted_Lead_Time",
            "Lead_Time_Reduction_Days",
            "Lead_Time_Reduction_Percent",
            "Speed_Score",
            "Scenario_Confidence_Score",
            "Recommendation_Score",
            "Recommendation_Rank",
            "Recommendation_Risk"
        ]
    ].round(2)
)

# ------------------------------------------------------------
# Save recommendations
# ------------------------------------------------------------

recommendation_output_path = (
    OUTPUT_DIR / "top_factory_reallocation_recommendations.csv"
)

top_recommendations.to_csv(
    recommendation_output_path,
    index=False
)

print("\nRecommendations saved to:")
print(recommendation_output_path)
```

```python
# ============================================================
# PROFIT STABILITY ANALYSIS
# ============================================================

# ------------------------------------------------------------
# Calculate historical profitability by product and region
# ------------------------------------------------------------

profit_stability = (
    clean_df
    .groupby(
        [
            "Product_ID",
            "Product_Name",
            "Region"
        ],
        as_index=False
    )
    .agg(
        Average_Profit_Margin=("Profit_Margin", "mean"),
        Profit_Margin_Std=("Profit_Margin", "std"),
        Average_Profit_Per_Unit=("Profit_Per_Unit", "mean"),
        Total_Gross_Profit=("Gross_Profit", "sum"),
        Records=("Row_ID", "count")
    )
)

# Replace zero/undefined standard deviation for
# single-observation groups
profit_stability["Profit_Margin_Std"] = (
    profit_stability["Profit_Margin_Std"]
    .fillna(0)
)

# ------------------------------------------------------------
# Create normalized profitability indicators
# ------------------------------------------------------------

margin_min = profit_stability["Average_Profit_Margin"].min()
margin_max = profit_stability["Average_Profit_Margin"].max()

variation_min = profit_stability["Profit_Margin_Std"].min()
variation_max = profit_stability["Profit_Margin_Std"].max()

if margin_max > margin_min:
    profit_stability["Margin_Score"] = (
        (
            profit_stability["Average_Profit_Margin"]
            - margin_min
        )
        / (margin_max - margin_min)
    ) * 100
else:
    profit_stability["Margin_Score"] = 100

if variation_max > variation_min:
    profit_stability["Stability_Score"] = (
        1
        - (
            (
                profit_stability["Profit_Margin_Std"]
                - variation_min
            )
            / (variation_max - variation_min)
        )
    ) * 100
else:
    profit_stability["Stability_Score"] = 100

# ------------------------------------------------------------
# Combined profit stability score
# ------------------------------------------------------------

profit_stability["Profit_Stability_Score"] = (
    0.60 * profit_stability["Margin_Score"]
    + 0.40 * profit_stability["Stability_Score"]
)

# Round values
round_columns = [
    "Average_Profit_Margin",
    "Profit_Margin_Std",
    "Average_Profit_Per_Unit",
    "Margin_Score",
    "Stability_Score",
    "Profit_Stability_Score"
]

for col in round_columns:
    profit_stability[col] = (
        profit_stability[col].round(2)
    )

print("=" * 70)
print("PROFIT STABILITY ANALYSIS")
print("=" * 70)

print("\nProduct-region profitability records:")
print(len(profit_stability))

display(
    profit_stability[
        [
            "Product_ID",
            "Product_Name",
            "Region",
            "Average_Profit_Margin",
            "Profit_Margin_Std",
            "Average_Profit_Per_Unit",
            "Profit_Stability_Score"
        ]
    ]
    .sort_values(
        "Profit_Stability_Score",
        ascending=False
    )
)

# ------------------------------------------------------------
# Save profit stability analysis
# ------------------------------------------------------------

profit_stability_path = (
    OUTPUT_DIR / "profit_stability_analysis.csv"
)

profit_stability.to_csv(
    profit_stability_path,
    index=False
)

print("\nProfit stability analysis saved to:")
print(profit_stability_path)
```

```python
# ============================================================
# COMBINED OPTIMIZATION SCORE
# ============================================================

# ------------------------------------------------------------
# Merge profit stability into recommendations
# ------------------------------------------------------------

optimization_candidates = refined_recommendations.merge(
    profit_stability[
        [
            "Product_ID",
            "Product_Name",
            "Region",
            "Average_Profit_Margin",
            "Profit_Stability_Score"
        ]
    ],
    on=[
        "Product_ID",
        "Product_Name",
        "Region"
    ],
    how="left"
)

# ------------------------------------------------------------
# Optimization priority
# ------------------------------------------------------------

# 0.70 means 70% priority to shipping speed
# and 30% priority to financial stability.
speed_priority = 0.70
profit_priority = 1 - speed_priority

# ------------------------------------------------------------
# Combined decision score
# ------------------------------------------------------------

optimization_candidates["Optimization_Score"] = (
    speed_priority
    * optimization_candidates["Speed_Score"]
    + profit_priority
    * optimization_candidates["Profit_Stability_Score"]
)

optimization_candidates["Optimization_Score"] = (
    optimization_candidates["Optimization_Score"]
    .round(2)
)

# ------------------------------------------------------------
# Risk classification
# ------------------------------------------------------------

optimization_candidates["Final_Risk_Level"] = np.select(
    [
        (
            optimization_candidates["Scenario_Confidence_Score"] >= 70
        )
        & (
            optimization_candidates["Profit_Stability_Score"] >= 70
        ),

        (
            optimization_candidates["Scenario_Confidence_Score"] >= 45
        )
        & (
            optimization_candidates["Profit_Stability_Score"] >= 50
        )
    ],
    [
        "Low",
        "Medium"
    ],
    default="High"
)

# ------------------------------------------------------------
# Final ranking within each product-region combination
# ------------------------------------------------------------

optimization_candidates["Final_Rank"] = (
    optimization_candidates
    .groupby(
        [
            "Product_ID",
            "Product_Name",
            "Region"
        ]
    )["Optimization_Score"]
    .rank(
        ascending=False,
        method="first"
    )
)

# ------------------------------------------------------------
# Final recommendation table
# ------------------------------------------------------------

final_recommendations = (
    optimization_candidates[
        optimization_candidates["Final_Rank"] <= 3
    ]
    .sort_values(
        [
            "Product_Name",
            "Region",
            "Final_Rank"
        ]
    )
    .reset_index(drop=True)
)

print("=" * 70)
print("COMBINED FACTORY OPTIMIZATION SCORE")
print("=" * 70)

print("\nSpeed priority:")
print(f"{speed_priority * 100:.0f}%")

print("Profit stability priority:")
print(f"{profit_priority * 100:.0f}%")

print("\nFinal recommendation records:")
print(len(final_recommendations))

print("\n" + "=" * 70)
print("FINAL RECOMMENDATION TABLE")
print("=" * 70)

display(
    final_recommendations[
        [
            "Product_ID",
            "Product_Name",
            "Region",
            "Candidate_Factory",
            "Current_Predicted_Lead_Time",
            "Predicted_Lead_Time",
            "Lead_Time_Reduction_Days",
            "Lead_Time_Reduction_Percent",
            "Scenario_Confidence_Score",
            "Average_Profit_Margin",
            "Profit_Stability_Score",
            "Optimization_Score",
            "Final_Rank",
            "Final_Risk_Level"
        ]
    ].round(2)
)

# ------------------------------------------------------------
# Save final recommendations
# ------------------------------------------------------------

final_recommendation_path = (
    OUTPUT_DIR / "final_factory_optimization_recommendations.csv"
)

final_recommendations.to_csv(
    final_recommendation_path,
    index=False
)

print("\nFinal recommendations saved to:")
print(final_recommendation_path)
```

```python
# ============================================================
# FINAL DASHBOARD-READY DECISION TABLE
# ============================================================

# ------------------------------------------------------------
# Get the best recommendation for each Product × Region
# ------------------------------------------------------------

best_recommendations = (
    final_recommendations[
        final_recommendations["Final_Rank"] == 1
    ][
        [
            "Product_ID",
            "Product_Name",
            "Region",
            "Candidate_Factory",
            "Predicted_Lead_Time",
            "Lead_Time_Reduction_Days",
            "Lead_Time_Reduction_Percent",
            "Scenario_Confidence_Score",
            "Profit_Stability_Score",
            "Optimization_Score",
            "Final_Risk_Level"
        ]
    ]
    .copy()
)

# ------------------------------------------------------------
# Current factory for each Product × Region
# ------------------------------------------------------------

current_assignment = (
    clean_df[
        [
            "Product_ID",
            "Product_Name",
            "Region",
            "Current_Factory"
        ]
    ]
    .drop_duplicates()
)

# ------------------------------------------------------------
# Product × Region transaction volume
# ------------------------------------------------------------

product_region_volume = (
    clean_df
    .groupby(
        [
            "Product_ID",
            "Product_Name",
            "Region"
        ],
        as_index=False
    )
    .agg(
        Records=("Row_ID", "count"),
        Total_Units=("Units", "sum"),
        Total_Sales=("Sales", "sum"),
        Total_Gross_Profit=("Gross_Profit", "sum")
    )
)

# ------------------------------------------------------------
# Build final decision table
# ------------------------------------------------------------

final_decision_table = (
    product_region_volume
    .merge(
        current_assignment,
        on=[
            "Product_ID",
            "Product_Name",
            "Region"
        ],
        how="left"
    )
    .merge(
        best_recommendations,
        on=[
            "Product_ID",
            "Product_Name",
            "Region"
        ],
        how="left"
    )
)

# ------------------------------------------------------------
# Create decision status
# ------------------------------------------------------------

final_decision_table["Decision_Status"] = np.where(
    final_decision_table["Candidate_Factory"].notna(),
    "Reallocation Opportunity",
    "Retain Current Assignment"
)

# ------------------------------------------------------------
# Fill non-recommendation fields
# ------------------------------------------------------------

final_decision_table["Recommended_Factory"] = (
    final_decision_table["Candidate_Factory"]
    .fillna(final_decision_table["Current_Factory"])
)

final_decision_table["Predicted_Lead_Time"] = (
    final_decision_table["Predicted_Lead_Time"]
    .fillna(
        final_decision_table["Current_Predicted_Lead_Time"]
        if "Current_Predicted_Lead_Time"
        in final_decision_table.columns
        else np.nan
    )
)

# For retained assignments, use current predicted lead time
if "Current_Predicted_Lead_Time" not in final_decision_table.columns:
    current_lead_time_data = (
        scenario_summary[
            scenario_summary["Is_Current_Factory"]
        ][
            [
                "Product_ID",
                "Product_Name",
                "Region",
                "Current_Predicted_Lead_Time"
            ]
        ]
        .drop_duplicates()
    )

    final_decision_table = (
        final_decision_table
        .merge(
            current_lead_time_data,
            on=[
                "Product_ID",
                "Product_Name",
                "Region"
            ],
            how="left"
        )
    )

    final_decision_table["Predicted_Lead_Time"] = (
        final_decision_table["Predicted_Lead_Time"]
        .fillna(
            final_decision_table[
                "Current_Predicted_Lead_Time"
            ]
        )
    )

# ------------------------------------------------------------
# Clean up recommendation columns
# ------------------------------------------------------------

final_decision_table["Lead_Time_Reduction_Days"] = (
    final_decision_table["Lead_Time_Reduction_Days"]
    .fillna(0)
)

final_decision_table["Lead_Time_Reduction_Percent"] = (
    final_decision_table["Lead_Time_Reduction_Percent"]
    .fillna(0)
)

final_decision_table["Scenario_Confidence_Score"] = (
    final_decision_table["Scenario_Confidence_Score"]
    .fillna(0)
)

final_decision_table["Profit_Stability_Score"] = (
    final_decision_table["Profit_Stability_Score"]
    .fillna(
        final_decision_table["Current_Factory"]
        .notna()
        .astype(int) * 0
    )
)

final_decision_table["Optimization_Score"] = (
    final_decision_table["Optimization_Score"]
    .fillna(0)
)

final_decision_table["Final_Risk_Level"] = (
    final_decision_table["Final_Risk_Level"]
    .fillna("Not Applicable")
)

# ------------------------------------------------------------
# Round numerical values
# ------------------------------------------------------------

round_columns = [
    "Total_Sales",
    "Total_Gross_Profit",
    "Predicted_Lead_Time",
    "Lead_Time_Reduction_Days",
    "Lead_Time_Reduction_Percent",
    "Scenario_Confidence_Score",
    "Profit_Stability_Score",
    "Optimization_Score"
]

for col in round_columns:
    if col in final_decision_table.columns:
        final_decision_table[col] = (
            final_decision_table[col].round(2)
        )

# ------------------------------------------------------------
# Sort recommendations first
# ------------------------------------------------------------

final_decision_table = (
    final_decision_table
    .sort_values(
        [
            "Decision_Status",
            "Lead_Time_Reduction_Days"
        ],
        ascending=[True, False]
    )
    .reset_index(drop=True)
)

print("=" * 70)
print("FINAL DASHBOARD-READY DECISION TABLE")
print("=" * 70)

print("\nTotal product-region combinations:")
print(len(final_decision_table))

print("\nReallocation opportunities:")
print(
    (
        final_decision_table["Decision_Status"]
        == "Reallocation Opportunity"
    ).sum()
)

print("\nRetained current assignments:")
print(
    (
        final_decision_table["Decision_Status"]
        == "Retain Current Assignment"
    ).sum()
)

print("\nSample decision table:")
display(final_decision_table.head(20))

# ------------------------------------------------------------
# Save dashboard-ready table
# ------------------------------------------------------------

decision_table_path = (
    OUTPUT_DIR / "dashboard_ready_decision_table.csv"
)

final_decision_table.to_csv(
    decision_table_path,
    index=False
)

print("\nDashboard-ready decision table saved to:")
print(decision_table_path)
```

```python
# ============================================================
# DASHBOARD SCENARIO MATRIX
# ============================================================

# ------------------------------------------------------------
# Aggregate predicted lead time for each dashboard scenario
# ------------------------------------------------------------

dashboard_scenarios = (
    factory_scenario_predictions
    .groupby(
        [
            "Product_ID",
            "Product_Name",
            "Division",
            "Region",
            "Ship_Mode",
            "Current_Factory",
            "Candidate_Factory",
            "Is_Current_Factory"
        ],
        as_index=False
    )
    .agg(
        Records=("Predicted_Lead_Time", "size"),
        Predicted_Lead_Time=("Predicted_Lead_Time", "mean")
    )
)

dashboard_scenarios["Predicted_Lead_Time"] = (
    dashboard_scenarios["Predicted_Lead_Time"].round(2)
)

# ------------------------------------------------------------
# Get current-factory prediction for each scenario
# ------------------------------------------------------------

current_dashboard_prediction = (
    dashboard_scenarios[
        dashboard_scenarios["Is_Current_Factory"]
    ][
        [
            "Product_ID",
            "Product_Name",
            "Region",
            "Ship_Mode",
            "Predicted_Lead_Time"
        ]
    ]
    .rename(
        columns={
            "Predicted_Lead_Time":
            "Current_Predicted_Lead_Time"
        }
    )
)

# ------------------------------------------------------------
# Add current baseline to all candidate factories
# ------------------------------------------------------------

dashboard_scenarios = dashboard_scenarios.merge(
    current_dashboard_prediction,
    on=[
        "Product_ID",
        "Product_Name",
        "Region",
        "Ship_Mode"
    ],
    how="left"
)

# ------------------------------------------------------------
# Calculate scenario improvement
# ------------------------------------------------------------

dashboard_scenarios["Lead_Time_Reduction_Days"] = (
    dashboard_scenarios["Current_Predicted_Lead_Time"]
    - dashboard_scenarios["Predicted_Lead_Time"]
)

dashboard_scenarios["Lead_Time_Reduction_Percent"] = np.where(
    dashboard_scenarios["Current_Predicted_Lead_Time"] != 0,
    (
        dashboard_scenarios["Lead_Time_Reduction_Days"]
        / dashboard_scenarios["Current_Predicted_Lead_Time"]
    ) * 100,
    0
)

dashboard_scenarios["Lead_Time_Reduction_Days"] = (
    dashboard_scenarios["Lead_Time_Reduction_Days"].round(2)
)

dashboard_scenarios["Lead_Time_Reduction_Percent"] = (
    dashboard_scenarios["Lead_Time_Reduction_Percent"].round(2)
)

# ------------------------------------------------------------
# Mark improvement opportunities
# ------------------------------------------------------------

dashboard_scenarios["Improvement_Available"] = (
    dashboard_scenarios["Lead_Time_Reduction_Days"] > 0
)

# ------------------------------------------------------------
# Sort for easy dashboard use
# ------------------------------------------------------------

dashboard_scenarios = (
    dashboard_scenarios
    .sort_values(
        [
            "Product_Name",
            "Region",
            "Ship_Mode",
            "Predicted_Lead_Time"
        ]
    )
    .reset_index(drop=True)
)

print("=" * 70)
print("DASHBOARD SCENARIO MATRIX")
print("=" * 70)

print("\nTotal dashboard scenarios:")
print(len(dashboard_scenarios))

print("\nUnique products:")
print(dashboard_scenarios["Product_Name"].nunique())

print("\nUnique regions:")
print(dashboard_scenarios["Region"].nunique())

print("\nUnique ship modes:")
print(dashboard_scenarios["Ship_Mode"].nunique())

print("\nUnique candidate factories:")
print(dashboard_scenarios["Candidate_Factory"].nunique())

print("\nSample dashboard scenarios:")
display(
    dashboard_scenarios.head(20)
)

# ------------------------------------------------------------
# Save dashboard scenario matrix
# ------------------------------------------------------------

dashboard_scenario_path = (
    OUTPUT_DIR / "dashboard_scenario_matrix.csv"
)

dashboard_scenarios.to_csv(
    dashboard_scenario_path,
    index=False
)

print("\nDashboard scenario matrix saved to:")
print(dashboard_scenario_path)
```

```python
# ============================================================
# FACTORY SENSITIVITY VALIDATION
# ============================================================

# ------------------------------------------------------------
# Measure variation in predicted lead time across factories
# for each Product × Region × Ship Mode combination
# ------------------------------------------------------------

factory_sensitivity = (
    dashboard_scenarios
    .groupby(
        [
            "Product_ID",
            "Product_Name",
            "Region",
            "Ship_Mode"
        ],
        as_index=False
    )
    .agg(
        Number_of_Factory_Options=(
            "Candidate_Factory",
            "nunique"
        ),
        Minimum_Predicted_Lead_Time=(
            "Predicted_Lead_Time",
            "min"
        ),
        Maximum_Predicted_Lead_Time=(
            "Predicted_Lead_Time",
            "max"
        ),
        Average_Predicted_Lead_Time=(
            "Predicted_Lead_Time",
            "mean"
        ),
        Factory_Prediction_Std=(
            "Predicted_Lead_Time",
            "std"
        )
    )
)

# ------------------------------------------------------------
# Calculate factory impact range
# ------------------------------------------------------------

factory_sensitivity["Factory_Impact_Range"] = (
    factory_sensitivity["Maximum_Predicted_Lead_Time"]
    - factory_sensitivity["Minimum_Predicted_Lead_Time"]
)

factory_sensitivity["Factory_Impact_Range"] = (
    factory_sensitivity["Factory_Impact_Range"].round(2)
)

factory_sensitivity["Factory_Prediction_Std"] = (
    factory_sensitivity["Factory_Prediction_Std"]
    .fillna(0)
    .round(2)
)

# ------------------------------------------------------------
# Count scenarios where factory changes prediction
# ------------------------------------------------------------

factory_sensitive_count = (
    factory_sensitivity["Factory_Impact_Range"] > 0.01
).sum()

factory_insensitive_count = (
    factory_sensitivity["Factory_Impact_Range"] <= 0.01
).sum()

total_sensitivity_groups = len(factory_sensitivity)

# ------------------------------------------------------------
# Display validation
# ------------------------------------------------------------

print("=" * 70)
print("FACTORY SENSITIVITY VALIDATION")
print("=" * 70)

print("\nTotal Product × Region × Ship Mode combinations:")
print(total_sensitivity_groups)

print("\nCombinations where factory changes predicted lead time:")
print(factory_sensitive_count)

print("\nCombinations with essentially identical predictions:")
print(factory_insensitive_count)

print("\nPercentage with factory-sensitive predictions:")

print(
    round(
        (
            factory_sensitive_count
            / total_sensitivity_groups
        ) * 100,
        2
    ),
    "%"
)

print("\n" + "=" * 70)
print("LARGEST FACTORY-IMPACT SCENARIOS")
print("=" * 70)

display(
    factory_sensitivity
    .sort_values(
        "Factory_Impact_Range",
        ascending=False
    )
    .head(20)
)

# ------------------------------------------------------------
# Save sensitivity analysis
# ------------------------------------------------------------

factory_sensitivity_path = (
    OUTPUT_DIR / "factory_sensitivity_analysis.csv"
)

factory_sensitivity.to_csv(
    factory_sensitivity_path,
    index=False
)

print("\nFactory sensitivity analysis saved to:")
print(factory_sensitivity_path)
```

## Project Limitations

Although the system provides a structured approach to factory reallocation analysis, several limitations should be considered while interpreting the results.

1. The dataset does not contain historical observations of every product being manufactured at every possible factory. Therefore, alternative factory results are model-based what-if predictions rather than directly observed outcomes.

2. The available shipment dates produce relatively large lead-time values. These values were retained as provided by the dataset and were not artificially adjusted during preprocessing.

3. Factory-specific transportation costs, production costs, factory capacity, inventory constraints, and distance-based logistics costs are not available. Therefore, the optimization process cannot calculate direct monetary savings from a factory reassignment.

4. The Scenario Confidence Score is a project-specific confidence proxy based on transaction volume and model performance. It is not a formal statistical confidence interval.

5. Profit Stability Score is based on available historical profit indicators. It does not represent a direct prediction of future profit after factory reassignment.

6. The selected Gradient Boosting model has strong predictive performance relative to the tested models, but its predictions still contain estimation error. Therefore, recommendations should be validated against real operational conditions before implementation.

These limitations define the current scope of the system and also indicate the types of additional data that could improve future versions of the optimization framework.
